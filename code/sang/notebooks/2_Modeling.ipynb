```python
import csv

# 1. Đọc dữ liệu từ file CSV bằng Python thuần
file_path = '../../../foody_sample_data.csv'
data = []

try:
    with open(file_path, mode='r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            data.append(row)
    print(f"Đọc thành công {len(data)} bản ghi từ file.")
except Exception as e:
    print(f"Lỗi khi đọc file: {e}")

# 2. Xử lý và lọc nhãn (Bỏ 3 sao, gán nhãn >=4 sao là 1, còn lại là 0)
dataset = []
for row in data:
    try:
        rating_val = float(row.get('rating', row.get('AvgRating', 0)))
        if rating_val == 3:
            continue 
        sentiment = 1 if rating_val >= 4 else 0
        text = row.get('review_text', '').lower()
        dataset.append({'text': text, 'label': sentiment})
    except ValueError:
        continue

print(f"Số lượng mẫu sau khi lọc nhãn: {len(dataset)}")

# 3. Định nghĩa từ khóa cho mô hình Baseline phân tích cảm xúc
positive_keywords = ['ngon', 'tuyệt', 'thích', 'nhanh', 'sạch', 'tốt', 'rẻ', 'đẹp', 'tươi', 'phục vụ tốt', 'đỉnh']
negative_keywords = ['dở', 'tệ', 'chán', 'lâu', 'đắt', 'bẩn', 'chậm', 'thái độ', 'kinh khủng', 'thất vọng']

def predict_sentiment(text):
    pos_score = sum(1 for word in positive_keywords if word in text)
    neg_score = sum(1 for word in negative_keywords if word in text)
    return 1 if pos_score >= neg_score else 0

# 4. Dự đoán và tính toán độ chính xác (Accuracy)
correct_predictions = 0
total_samples = len(dataset)

for item in dataset:
    pred = predict_sentiment(item['text'])
    if pred == item['label']:
        correct_predictions += 1

accuracy = (correct_predictions / total_samples) * 100 if total_samples > 0 else 0

# 5. In kết quả cuối cùng ra màn hình
print("\n--- KẾT QUẢ ĐÁNH GIÁ MÔ HÌNH BASELINE ---")
print(f"Tổng số mẫu kiểm tra: {total_samples}")
print(f"Số mẫu dự đoán đúng: {correct_predictions}")
print(f"Độ chính xác (Accuracy): {accuracy:.2f}%")
```
