```python
import csv

# Đường dẫn tới file dữ liệu mẫu
file_path = '../../../foody_sample_data.csv'
data = []

# Đọc file CSV bằng thư viện chuẩn 'csv'
try:
    with open(file_path, mode='r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            data.append(row)
    print(f"Đọc thành công tổng cộng {len(data)} dòng dữ liệu!")
except Exception as e:
    print(f"Lỗi khi đọc file: {e}")

# Xem 5 dòng đầu tiên của dữ liệu
print("\n--- 5 DÒNG DỮ LIỆU ĐẦU TIÊN ---")
for idx, row in enumerate(data[:5]):
    print(f"Dòng {idx + 1}: {row}")

# Kiểm tra thông tin tổng quan (Số dòng và các cột)
if data:
    print(f"\n--- THÔNG TIN TỔNG QUAN ---")
    print(f"Tổng số bản ghi: {len(data)}")
    print(f"Các cột có trong dữ liệu: {list(data[0].keys())}")
```
